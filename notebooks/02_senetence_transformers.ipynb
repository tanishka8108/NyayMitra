```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
print("Model loaded successfully!")
```

```python
sentence = "My neighbour built a wall on my land without permission"
embedding = model.encode(sentence)

print("Embedding shape:", embedding.shape)
print("First 10 numbers:", embedding[:10])
```

```python
from sentence_transformers import util

sentences = [
    "My neighbour built a wall on my land without permission",
    "Someone encroached on my property",
    "I want to order a pizza",
]

embeddings = model.encode(sentences)

similarity_1_2 = util.cos_sim(embeddings[0], embeddings[1])
print("Similarity (land dispute vs encroachment):", similarity_1_2.item())

similarity_1_3 = util.cos_sim(embeddings[0], embeddings[2])
print("Similarity (land dispute vs pizza):", similarity_1_3.item())
```

```python
law_descriptions = [
    "Someone deceived you to hand over money or property",
    "A cheque you received bounced due to insufficient funds",
    "Someone entered or used your property without permission",
    "Someone physically hurt you",
]

law_embeddings = model.encode(law_descriptions)

query = "my neighbor built a wall on my land"
query_embedding = model.encode(query)

similarities = util.cos_sim(query_embedding, law_embeddings)
print(similarities)
```

```python
best_match_idx = similarities.argmax().item()
best_score = similarities[0][best_match_idx].item()

print(f"Best match: '{law_descriptions[best_match_idx]}'")
print(f"Score: {best_score:.3f}")
```

```python
test_queries = [
    "I lent someone money and they never paid me back",
    "my business partner cheated me out of profits",
    "someone slapped me during a fight",
    "asdkjhaskjdh random gibberish text",
]

for q in test_queries:
    q_emb = model.encode(q)
    sims = util.cos_sim(q_emb, law_embeddings)
    best_idx = sims.argmax().item()
    best_score = sims[0][best_idx].item()
    print(f"\nQuery: {q}")
    print(f"Best match: {law_descriptions[best_idx]} (score: {best_score:.3f})")
```
